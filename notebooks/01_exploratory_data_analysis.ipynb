# ⚡ Energy Consumption Forecasting — Exploratory Data Analysis

This notebook explores the electrical load patterns, seasonal behavior, temperature dependencies, and production shift characteristics of the facility dataset.

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Resolve project paths
ROOT_DIR = Path("..").resolve()
PROCESSED_DATA_PATH = ROOT_DIR / "data" / "processed" / "cleaned_facility_load.csv"

print(f"Target processed data path: {PROCESSED_DATA_PATH}")
```

```python
# Load cleaned facility consumption data
df = pd.read_csv(PROCESSED_DATA_PATH)
df["timestamp"] = pd.to_datetime(df["timestamp"])
df = df.set_index("timestamp")
df.info()
```

```python
# Display summary statistics of load and weather
df[["load_kw", "temperature_c", "production_status"]].describe()
```

```python
# Plot sample 2-week load profile
sample_slice = df.iloc[-336:]  # Last 14 days
plt.figure(figsize=(14, 5))
plt.plot(sample_slice.index, sample_slice["load_kw"], label="Load (kW)", color="#1f77b4")
plt.title("Facility Electrical Load Profile (14-Day Window)")
plt.xlabel("Timestamp")
plt.ylabel("Load (kW)")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
# Hourly diurnal pattern by production shift status
df["hour"] = df.index.hour
hourly_summary = df.groupby(["hour", "production_status"])["load_kw"].mean().unstack()

plt.figure(figsize=(10, 5))
plt.plot(hourly_summary.index, hourly_summary[0], marker="o", label="Standby / Off Shift (Status=0)")
plt.plot(hourly_summary.index, hourly_summary[1], marker="s", label="Active Production (Status=1)")
plt.title("Average Load Profile Across 24 Hours by Shift Status")
plt.xlabel("Hour of Day (0-23)")
plt.ylabel("Mean Load (kW)")
plt.xticks(range(0, 24))
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
# Relationship between ambient temperature and power demand
plt.figure(figsize=(8, 6))
plt.scatter(df["temperature_c"], df["load_kw"], alpha=0.15, color="#ff7f0e", s=10)
plt.title("Electrical Load vs. Ambient Outdoor Temperature")
plt.xlabel("Ambient Temperature (°C)")
plt.ylabel("Electricity Load (kW)")
plt.grid(True, alpha=0.3)
plt.show()
```
